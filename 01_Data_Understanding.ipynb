Fix package_weight

```python
def clean_package_weight(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # Missing / invalid values
    if value.lower() in ["unknown", "-", ""]:
        return np.nan

    # Values explicitly in grams
    if value.lower().endswith("g") and not value.lower().endswith("kg"):
        try:
            weight = float(value[:-1].strip()) / 1000
            return weight if weight >= 0 else np.nan
        except:
            return np.nan

    # Values explicitly in kilograms
    if value.lower().endswith("kg"):
        try:
            weight = float(value[:-2].strip())
            return weight if weight >= 0 else np.nan
        except:
            return np.nan

    # Values without a unit → assume kg
    try:
        weight = float(value)
        return weight if weight >= 0 else np.nan
    except:
        return np.nan


orders["package_weight"] = orders["package_weight"].apply(clean_package_weight)
```

```python
import numpy as np

def clean_package_weight(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value.lower() in ['unknown', '-', '', 'nan']:
        return np.nan

    if value.lower().endswith('kg'):
        try:
            weight = float(value[:-2].strip())
            return weight if weight >= 0 else np.nan
        except:
            return np.nan

    if value.lower().endswith('g'):
        try:
            weight = float(value[:-1].strip()) / 1000
            return weight if weight >= 0 else np.nan
        except:
            return np.nan

    try:
        weight = float(value)
        return weight if weight >= 0 else np.nan
    except:
        return np.nan
```

```python
def clean_capacity(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().lower()

    # kilograms — check BEFORE grams
    if value.endswith('kg'):
        return float(value[:-2].strip())

    # grams → kilograms
    if value.endswith('g'):
        return float(value[:-1].strip()) / 1000

    # plain numeric value
    return float(value)


fleet_vehicles['capacity_kg'] = fleet_vehicles['capacity_kg'].apply(
    clean_capacity
)

```

```python
weather_data['temp_celsius'] = weather_data['temp']

# Convert Fahrenheit to Celsius where the unit is F
f_mask = weather_data['Temp Unit'].eq('F')

weather_data.loc[f_mask, 'temp_celsius'] = (
    weather_data.loc[f_mask, 'temp'] - 32
) * 5 / 9

# Keep only temperatures with a known unit
unknown_mask = weather_data['Temp Unit'].isna()
weather_data.loc[unknown_mask, 'temp_celsius'] = pd.NA

print(weather_data[['temp', 'Temp Unit', 'temp_celsius']].head(20))
print()
print("Missing standardized temperatures:",
      weather_data['temp_celsius'].isna().sum())
```
